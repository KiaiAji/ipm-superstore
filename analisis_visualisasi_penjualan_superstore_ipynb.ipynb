```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(
    "Sample - Superstore.csv",
    encoding="latin1"
)

df.head()
```

```python
print("Jumlah baris dan kolom:", df.shape)
print("\nNama kolom:")
print(df.columns.tolist())

print("\nInformasi dataset:")
df.info()
```

```python
print("Jumlah data kosong setiap kolom:")
print(df.isnull().sum())
```

```python
penjualan = df.groupby("Category")["Sales"].sum()

print("Total penjualan per kategori:")
print(penjualan.sort_values(ascending=False))
```

```python
penjualan.sort_values().plot(
    kind="bar",
    title="Total Penjualan per Kategori Produk"
)

plt.xlabel("Kategori Produk")
plt.ylabel("Total Penjualan")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
keuntungan = df.groupby("Category")["Profit"].sum()

print("Total keuntungan per kategori:")
print(keuntungan.sort_values(ascending=False))
```

```python
keuntungan.sort_values().plot(
    kind="bar",
    title="Total Keuntungan per Kategori Produk"
)

plt.xlabel("Kategori Produk")
plt.ylabel("Total Keuntungan")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
kategori_terlaris = penjualan.idxmax()
kategori_untung = keuntungan.idxmax()

print("Kategori dengan penjualan tertinggi:",
      kategori_terlaris)

print("Kategori dengan keuntungan tertinggi:",
      kategori_untung)
```
